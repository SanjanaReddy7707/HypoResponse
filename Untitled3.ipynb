```python

import json
import pandas as pd

with open("results/pilot.jsonl") as f:
    pilot = [json.loads(line) for line in f if line.strip()]

df = pd.DataFrame([
    {
        "System": r["system"],
        "Scenario": r["scenario_id"],
        "Tokens": r["metrics"]["tokens"],
        "LLM Calls": r["metrics"]["llm_calls"]
    }
    for r in pilot
])

display(df)

print("Total tokens:", df["Tokens"].sum())
print("Average tokens per job:", df["Tokens"].mean())
print("Total API calls:", df["LLM Calls"].sum())
```

```python

avg_tokens = df["Tokens"].mean()

main_jobs = 30 * 6 * 2
estimated_tokens = avg_tokens * main_jobs

print("Main evaluation jobs:", main_jobs)
print("Estimated total tokens:", round(estimated_tokens))
```

```python

import os
import shutil

backup = "/content/drive/MyDrive/HypoResponse_Results"

os.makedirs(backup, exist_ok=True)

shutil.copy(
    "results/pilot.jsonl",
    os.path.join(backup, "pilot.jsonl")
)

print("Backup folder:", backup)
```

```python

from pathlib import Path

local_raw = Path("/content/hyporesponse/results/raw.jsonl")
drive_raw = Path(backup) / "raw.jsonl"

if local_raw.is_symlink():
    local_raw.unlink()
elif local_raw.exists():
    shutil.copy2(local_raw, drive_raw)
    local_raw.unlink()

local_raw.symlink_to(drive_raw)

print("Main evaluation results will save to:", drive_raw)
```

```python

import json
from collections import Counter

with open("results/raw.jsonl") as f:
    results = [json.loads(line) for line in f if line.strip()]

unique_jobs = {
    (r["system"], r["scenario_id"], r["run"])
    for r in results
}

print("Expected jobs:", 360)
print("Completed unique jobs:", len(unique_jobs))
print("Missing jobs:", 360 - len(unique_jobs))
print("System counts:", Counter(r["system"] for r in results))
```
